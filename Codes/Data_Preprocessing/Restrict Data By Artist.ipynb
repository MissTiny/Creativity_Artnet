```python
artist=[]
print(f"{datetime.now():%Y-%m-%d %H:%M:%S}: Start")
for i in range(artnet_2025.shape[0]):
    row = artnet_2025.iloc[i]
    if row['first']:
        artist.append(f"{row['first']} {row['last']}")
    else:
        artist.append(f"{row['last']}")
    if i%10000==0:
        print(f"{datetime.now():%Y-%m-%d %H:%M:%S}: Currently at {i}")
print(f"{datetime.now():%Y-%m-%d %H:%M:%S}: End")
    
```
